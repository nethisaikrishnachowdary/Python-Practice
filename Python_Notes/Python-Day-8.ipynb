What are Instance Variables?
In Python Object-Oriented Programming (OOP), instance variables are variables that are bound to a specific instance (object) of a class. Unlike class variables which are shared across all instances, instance variables are unique to each individual object.

Key Characteristics:
Scope & Ownership: They belong to the specific object creation. If you change an instance variable for one object, it has zero impact on other objects of the same class.

The self Keyword: They are defined and accessed inside class methods using self (e.g., self.variable_name). The self parameter represents the specific object you are currently creating or manipulating.

Initialization: They are typically defined inside the constructor method, __init__, which automatically runs when you instantiate a new object.

```python
class SmartDevice:
    # Constructor method to initialize instance variables
    def __init__(self, device_type, brand, battery_level):
        # These are all instance variables
        self.device_type = device_type
        self.brand = brand
        self.battery_level = battery_level

    # Method to display the unique status of the instance
    def check_status(self):
        return f"{self.brand} {self.device_type}: {self.battery_level}% battery remaining."

# --- Creating separate instances (objects) ---
phone = SmartDevice(device_type="Smartphone", brand="Apple", battery_level=85)
laptop = SmartDevice(device_type="Laptop", brand="Dell", battery_level=42)

# 1. Accessing unique instance variables
print(phone.check_status())   # Output: Apple Smartphone: 85% battery remaining.
print(laptop.check_status())  # Output: Dell Laptop: 42% battery remaining.

# 2. Modifying an instance variable on a single object
phone.battery_level = 100

print("\n--- After charging the phone ---")
print(phone.check_status())   # Output: Apple Smartphone: 100% battery remaining.
print(laptop.check_status())  # Output: Dell Laptop: 42% battery remaining. (Remains unchanged)
```

In Python, a dataclass is a specialized class designed primarily to store data. Introduced in Python 3.7 via the dataclasses module, it uses a decorator (@dataclass) to automatically generate repetitive boilerplate code (dunder methods) that you would normally have to write by hand.

When you decorate a class with @dataclass, Python automatically writes the following methods for you behind the scenes:

__init__(): Handles object initialization and assigns arguments to instance variables.

__repr__(): Generates a clean, readable string representation of the object (e.g., Product(name='Laptop', price=999.99) instead of <__main__.Product object at 0x...>).

__eq__(): Allows you to compare two objects by their values (obj1 == obj2) rather than their memory addresses.

Why Use Data Classes?
Less Boilerplate: You don't have to write repetitive self.x = x assignments.

Readability: Type hints are required, making the data structures explicit and easy to read.

Built-in Functionality: Easily supports features like immutability (read-only objects) and ordering out of the box.

```python
from dataclasses import dataclass, field

# 1. Defining a basic Data Class
@dataclass
class InvestmentAsset:
    ticker: str
    asset_class: str
    current_price: float
    units_held: float
    # You can set default values directly
    is_active: bool = True 

# 2. Creating instances (The __init__ method is automatically generated)
asset1 = InvestmentAsset("AAPL", "Equity", 185.50, 50.0)
asset2 = InvestmentAsset("AAPL", "Equity", 185.50, 50.0)
asset3 = InvestmentAsset("BTC", "Crypto", 65000.0, 0.5)

# 3. Automatic __repr__ (Prints a beautifully formatted string)
print("--- Automatic __repr__ ---")
print(asset1)  
# Output: InvestmentAsset(ticker='AAPL', asset_class='Equity', current_price=185.5, units_held=50.0, is_active=True)

# 4. Automatic __eq__ (Compares values, not memory addresses)
print("\n--- Automatic __eq__ Comparison ---")
print(f"Is asset1 equal to asset2? {asset1 == asset2}")  # Output: True
print(f"Is asset1 equal to asset3? {asset1 == asset3}")  # Output: False

# 5. Adding a regular method to a Data Class
@dataclass
class PortfolioValue:
    ticker: str
    price: float
    quantity: float

    # You can still write custom methods just like a regular class
    def get_total_value(self) -> float:
        return self.price * self.quantity

stock = PortfolioValue("MSFT", 420.00, 10)
print("\n--- Custom Method Execution ---")
print(f"Total {stock.ticker} Value: ${stock.get_total_value():,.2f}")
# Output: Total MSFT Value: $4,200.00
```




Polymorphism is one of the core pillars of Object-Oriented Programming (OOP). The word originates from Greek, meaning "many forms" (poly = many, morph = form). In programming, it refers to the ability of different objects to respond to the same method call in their own unique way.

Instead of writing separate functions for every single object type, polymorphism allows you to write clean, flexible code that can treat different objects as if they belong to a common superclass, while executing the behavior specific to each subclass.

Types of Polymorphism in Python:
Duck Typing (Dynamic Polymorphism): Python's philosophy is "If it walks like a duck and quacks like a duck, it's a duck." If an object has a specific method, Python will execute it, regardless of the object's actual class hierarchy.

Method Overriding (Runtime Polymorphism): A child class provides a specific implementation of a method that is already defined in its parent class.

```python
class SQLDatabase:
    """Parent class representing a generic database connection."""
    def connect(self) -> str:
        raise NotImplementedError("Subclasses must implement this method")

class SQLiteConnection(SQLDatabase):
    """Subclass implementing a local SQLite connection."""
    def __init__(self, db_path: str):
        self.db_path = db_path

    def connect(self) -> str:
        return f"Connected to local SQLite database at: {self.db_path}"

class PostgreSQLConnection(SQLDatabase):
    """Subclass implementing a remote PostgreSQL connection."""
    def __init__(self, host: str, port: int):
        self.host = host
        self.port = port

    def connect(self) -> str:
        return f"Connected to remote PostgreSQL cluster at {self.host}:{self.port}"

# --- Polymorphism in Action ---

# A single function that handles any database instance uniformly
def initialize_database(db_instance: SQLDatabase):
    # This function doesn't care if it's SQLite or PostgreSQL.
    # It just calls .connect() and trusts the object to handle it.
    print("Initializing system...")
    connection_status = db_instance.connect()
    print(connection_status)
    print("Initialization complete.\n")

# Creating different database objects
local_db = SQLiteConnection("data/investment_copilot.db")
cloud_db = PostgreSQLConnection("192.168.1.50", 5432)

# Passing different objects into the exact same function
initialize_database(local_db)
initialize_database(cloud_db)
```

Encapsulation is another core pillar of Object-Oriented Programming (OOP) that involves bundling data (variables) and the methods that operate on that data into a single unit (a class).

More importantly, encapsulation is about data hiding. It restricts direct access to some of an object's components, which prevents accidental modification or misuse of the internal state. Instead of allowing external code to tamper with variables directly, the class provides public methods (called getters and setters) to safely inspect and modify the data.

Access Modifiers in Python
Unlike languages like Java or C++, Python does not have strict private keywords. Instead, it relies on naming conventions to signal how a variable should be treated:

Public: Accessible from anywhere. (e.g., self.variable)

Protected: A signal to other developers that the variable is intended for internal use within the class and its subclasses. Python won't strictly block access. (e.g., self._variable)

Private: Triggers Name Mangling. Python changes the internal name of the attribute behind the scenes, making it much harder to access or modify accidentally from outside the class. (e.g., self.__variable)

```python


class BankAccount:
    def __init__(self, owner: str, initial_balance: float):
        self.owner = owner                # Public attribute
        self.__balance = initial_balance  # Private attribute (hidden from direct access)

    # Getter method: Allows controlled access to read the private data
    def get_balance(self) -> float:
        return self.__balance

    # Setter method: Allows controlled modification with validation rules
    def deposit(self, amount: float) -> None:
        if amount > 0:
            self.__balance += amount
            print(f"Successfully deposited ${amount:,.2f}")
        else:
            print("Error: Deposit amount must be positive.")

    # Setter method: Encapsulates the logic and safety checks for changing data
    def withdraw(self, amount: float) -> None:
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Successfully withdrew ${amount:,.2f}")
        else:
            print("Error: Insufficient funds or invalid amount.")

# --- Testing Encapsulation ---

account = BankAccount("Saikrishna", 1000.00)

# 1. Accessing public data works fine
print(f"Account Owner: {account.owner}") # Output: Saikrishna

# 2. Attempting to access private data directly throws an AttributeError
try:
    print(account.__balance)
    
except AttributeError as e:
    print(f"\nDirect Access Blocked: {e}") 
    # Output: 'BankAccount' object has no attribute '__balance'

# 3. Accessing and modifying data safely through public methods
print(f"\nInitial Balance via Getter: ${account.get_balance():,.2f}")

account.deposit(500.00)     # Updates balance internally
account.withdraw(200.00)    # Updates balance internally

print(f"Final Balance via Getter: ${account.get_balance():,.2f}")

# 4. The setter protects the data from bad logic
account.withdraw(5000.00)   # Output: Error: Insufficient funds or invalid amount.
```
