File Handling in Python:
File handling refers to the process of performing operations on a file, such as creating, opening, reading, writing and closing it through a programming interface. It involves managing the data flow between the program and the file system on the storage device, ensuring that data is handled safely and efficiently.

Need for File Handling
Store data permanently, even after the program ends.
Access external files like .txt, .csv, .json, etc.
Process large files efficiently without using much memory.
Automate tasks like reading configs or saving outputs.

"r" - Read - Default value. Opens a file for reading, error if the file does not exist
"a" - Append - Opens a file for appending, creates the file if it does not exist
"w" - Write - Opens a file for writing, creates the file if it does not exist
"x" - Create - Creates the specified file, returns an error if the file exists
"t" - Text - Default value. Text mode
"b" - Binary - Binary mode (e.g. images)

```python
# 1. Navigating Inside a File: tell() and seek()
# When you open a file, think of the computer as keeping its place with an invisible cursor or pointer. As you read or write, that cursor moves forward.

# tell()
# This method returns the current position of the file pointer, measured in bytes from the beginning of the file.

# seek(offset, whence)
# This method allows you to move the cursor to a specific position in the file.
# offset is how many bytes to move.
# whence (optional) is where to start from: 0 (default, beginning of file), 1 (current position), or 2 (end of file).
with open("secret_code.txt", "r") as file:
    print(file.tell())    # Outputs: 0 (we just opened it)
    
    file.read(5)          # Read the first 5 characters
    print(file.tell())    # Outputs: 5
    
    file.seek(0)          # Jump back to the very beginning!
    print(file.tell())    # Outputs: 0


# 2. Managing File Storage: flush() and truncate()
# flush()
# When you write data to a file using write(), the computer doesn't always save it to the hard drive immediately. Instead, it holds it in a temporary memory buffer because writing to hardware is slow.

# Calling flush() forces the computer to empty that buffer and push everything onto the disk right now, without closing the file. (Note: close() calls flush() automatically behind the scenes).

# truncate(size)
# This method resizes the file to a specified number of bytes. If you don't provide a size, it cuts the file off right where the cursor currently is, deleting everything after it.

with open("log.txt", "r+") as file:  # "r+" allows both reading and writing
    file.truncate(10)                # The file is now exactly 10 bytes long. 
                                     # Everything else is erased!

# 3. Investigating File States (Boolean Methods)Sometimes you need your code to double-check the state of a file before trying to perform an action. 
# These methods all return True or False.
# readable(),"Returns True if the file can be read (e.g., it wasn't opened strictly in write 'w' mode)."
# writable(),Returns True if the file can be written to.
# seekable(),"Returns True if the file allows cursor navigation (some specialized files, like live stream inputs, don't allow you to jump around)."
# closed,"(This is an attribute, not a method, so no parentheses). Returns True if the file is closed."

file = open("data.txt", "w")
print(file.readable())  # Outputs: False (because we opened it in 'w' mode)
print(file.writable())  # Outputs: True
file.close()

print(file.closed)      # Outputs: True
```

```python
# Check if File exist:
# To avoid getting an error, you might want to check if the file exists before you try to delete it:
# To delete a file, you must import the OS module, and run its os.remove() function:

import os
if os.path.exists("demofile.txt"):
  os.remove("demofile.txt")
else:
  print("The file does not exist")
    
# Delete Folder:
# To delete an entire folder, use the os.rmdir() method:
# Remove the folder "myfolder":

import os
os.rmdir("myfolder")
```

```python
# Write to an Existing File
# To write to an existing file, you must add a parameter to the open() function:
# "a" - Append - will append to the end of the file
# "w" - Write - will overwrite any existing content
with open("demofile.txt", "a") as f:
  f.write("Now the file has more content!")

#open and read the file after the appending:
with open("demofile.txt") as f:
  print(f.read())
# To overwrite the existing content to the file, use the w parameter:
with open("demofile.txt", "w") as f:
  f.write("Woops! I have deleted the content!")

#open and read the file after the overwriting:
with open("demofile.txt") as f:
  print(f.read())

# Create a new file called "myfile.txt":
# To create a new file in Python, use the open() method, with one of the following parameters:
# "x" - Create - will create a file, returns an error if the file exists
# "a" - Append - will create a file if the specified file does not exists
# "w" - Write - will create a file if the specified file does not exists
f = open("myfile.txt", "x")
```
