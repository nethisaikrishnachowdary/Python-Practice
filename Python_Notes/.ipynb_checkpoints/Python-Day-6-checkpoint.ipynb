## 🔍 Pandas Filtering Cheat Sheet

| Filter Type | Syntax Example | Quick One-Line Description |
| :--- | :--- | :--- |
| **Equal To** | `df[df['Col'] == 'Val']` | Keeps rows where the column exactly matches a specific value. |
| **Not Equal To** | `df[df['Col'] != 'Val']` | Keeps rows where the column does not match a specific value. |
| **Greater Than** | `df[df['Col'] > 50]` | Keeps rows where a numerical column is strictly above a threshold. |
| **Less Than** | `df[df['Col'] < 50]` | Keeps rows where a numerical column is strictly below a threshold. |
| **Greater or Equal** | `df[df['Col'] >= 50]` | Keeps rows where a numerical column is above or exactly meets a value. |
| **Less or Equal** | `df[df['Col'] <= 50]` | Keeps rows where a numerical column is below or exactly meets a value. |
| **Logical AND** | `df[(df['A'] > 5) & (df['B'] < 10)]` | Keeps rows only when **all** specified conditions are simultaneously true. |
| **Logical OR** | `df[(df['A'] > 5) \| (df['B'] < 10)]` | Keeps rows if **at least one** of the specified conditions is true. |
| **Logical NOT** | `df[~(df['Col'] == 'Val')]` | Inverts the condition inside, keeping everything that fails the rule. |
| **List Matching** | `df[df['Col'].isin(['A', 'B'])]` | Keeps rows where the column value matches any item in a provided list. |
| **Range Filter** | `df[df['Col'].between(10, 20)]` | Keeps rows where numerical values fall inside a specific inclusive range. |
| **Find Missing** | `df[df['Col'].isna()]` | Keeps rows where the cell is empty, blank, or contains a NaN value. |
| **Find Filled** | `df[df['Col'].notna()]` | Drops empty rows and keeps only rows containing valid, existing data. |
| **Text Contains** | `df[df['Col'].str.contains('abc')]` | Keeps rows where a text column contains a specific partial substring. |
| **Text Starts With** | `df[df['Col'].str.startswith('A')]` | Keeps rows where a text column begins with a designated character/string. |
| **Text Ends With** | `df[df['Col'].str.endswith('Z')]` | Keeps rows where a text column terminates with a designated character/string. |
| **SQL-Style String** | `df.query("Col == 'Val'")` | Filters the table using a clean, readable text string instead of brackets. |

```python
# df.head()
df['Price']=df['Price'].str.replace(',','').str.replace('$','').str.replace('Ask For Price','')
df['Price']=pd.to_numeric(df['Price'],errors='coerce')
```

```python
df
```

```python
df['kms_driven']=df['kms_driven'].str.replace(',','').str.replace('kms','')
df['kms_driven']=pd.to_numeric(df['kms_driven'],errors='coerce')
```

```python
df['year']=pd.to_numeric(df['year'],errors='coerce')
```

```python
df=df.dropna(subset=['Price', 'year', 'kms_driven', 'fuel_type'])
```

## 🎨 1. Aesthetic Control (Styling the Canvas)
| Keyword | One Line Description | Example |
| :--- | :--- | :--- |
| `figsize` | Sets the dimensions of the figure (width, height) in inches. | `plt.figure(figsize=(10, 5))` |
| `color` | Defines the specific color for lines, bars, or scatter points. | `plt.plot(x, y, color='red')` |
| `alpha` | Sets the transparency level of elements (0.0 to 1.0). | `plt.scatter(x, y, alpha=0.5)` |
| `linestyle` | Determines the style of lines (solid, dashed, dotted). | `plt.plot(x, y, linestyle='--')` |

---

## 🏷️ 2. Labeling & Context (Making Charts Readable)
| Keyword | One Line Description | Example |
| :--- | :--- | :--- |
| `title()` | Adds a descriptive main heading to the top of the plot. | `plt.title("Price vs. Kms")` |
| `xlabel()` | Sets the label text for the horizontal axis. | `plt.xlabel("Kilometers")` |
| `ylabel()` | Sets the label text for the vertical axis. | `plt.ylabel("Price")` |
| `legend()` | Displays a key identifying the different data series. | `plt.legend(["Used Cars"])` |

---

## 📈 3. Axis & Grid Controls (The Viewing Window)
| Keyword | One Line Description | Example |
| :--- | :--- | :--- |
| `xlim()` | Sets the range displayed on the X-axis (min, max). | `plt.xlim(0, 100000)` |
| `ylim()` | Sets the range displayed on the Y-axis (min, max). | `plt.ylim(0, 500000)` |
| `grid()` | Adds a background grid to make reading values easier. | `plt.grid(True, alpha=0.3)` |
| `xticks()` | Manually defines which tick marks appear on the X-axis. | `plt.xticks([0, 50, 100])` |

---

## 🚀 4. Seaborn Plot Keywords (Statistical Mapping)
| Keyword | One Line Description | Example |
| :--- | :--- | :--- |
| `data` | Specifies the DataFrame containing the columns to plot. | `sns.barplot(data=df, ...)` |
| `x` | Defines the column or data used for the horizontal axis. | `sns.scatterplot(x='year', ...)` |
| `y` | Defines the column or data used for the vertical axis. | `sns.scatterplot(y='Price', ...)` |
| `hue` | Adds a third dimension by coloring points based on a category. | `sns.scatterplot(..., hue='fuel_type')` |

Chart 1: Scatter Plot (Kms Driven vs. Price)
Goal: See how a car's mileage affects its resale value. We expect a downward trend (higher kilometers = lower price).


```python
plt.figure(figsize=(5,5))
sns.scatterplot(data=df,x='kms_driven',y='Price',alpha=0.5,color='black')
plt.title("Kilometers driven vs price")
plt.xlabel("Kms diven")
plt.ylabel("Price")
plt.grid(True,linestyle='--',alpha=0.6)
plt.show()
```

Chart 2: Box Plot / Bar Plot (Fuel Type vs. Price)
Goal: Compare categorical data against numerical data. Let's see if Diesel, Petrol, or Electric cars hold their value better on average.

```python
plt.figure(figsize=(5,5))
sns.barplot(data=df,x='fuel_type',y='Price',alpha=0.5,color='darkred')
plt.title("fuel type vs price")
plt.xlabel("fuel")
plt.ylabel("Price")
plt.grid(True,linestyle='--',alpha=0.6)
plt.show()
```

Chart 3: Trend Line Plot (Year vs. Price)
Goal: Track how prices change based on the car's manufacturing year. We'll use Seaborn's lineplot which handles multiple data points per year by drawing a trend line and a shaded confidence interval.

```python
plt.figure(figsize=(5,5))
sns.lineplot(data=df,x='year',y='Price',alpha=0.5,color='darkred')
plt.title("year vs price")
plt.xlabel("year")
plt.ylabel("Price")
plt.grid(True,linestyle='--',alpha=0.6)
plt.show()
```

```python
import plotly.express as px

# Create an interactive 3D scatter plot using Plotly Express
fig = px.scatter_3d(
    df, 
    x='kms_driven', 
    y='Price', 
    z='year',
    color='fuel_type',                  # Automatically color-codes your points safely!
    title='Interactive 3D Used Car Analysis',
    labels={
        'kms_driven': 'Kilometers Driven',
        'Price': 'Price',
        'year': 'Manufacturing Year'
    }
)

# Update the visual size of the dots so they look clean
fig.update_traces(marker=dict(size=5))

# Display the interactive chart in your Notebook
fig.show()
```
