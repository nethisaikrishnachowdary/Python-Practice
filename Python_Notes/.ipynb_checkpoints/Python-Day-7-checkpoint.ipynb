Python Classes and Objects
Before looking at code, let's look at the philosophy. In standard procedural programming, you focus on actions (functions). In Object-Oriented Programming (OOP), you focus on things (objects).

Think of a Class as a blueprint on a piece of paper. It defines what a thing is and what it can do, but the blueprint itself isn't a physical thing. An Object (also called an Instance) is the actual, physical thing built from that blueprint.

The Real-World Analogy
Imagine an architect draws a blueprint for a house.

The Class is the blueprint. It says every house has a door color, a number of bedrooms, and a way to lock the front door.

The Objects are the actual houses built on the street. You can build a blue house with 3 bedrooms, and a yellow house with 4 bedrooms. They both used the same blueprint, but they are separate, unique houses.

```python
# 1. Define the Blueprint (The Class)
# By convention, class names always start with a Capital Letter.
class House:
    # This is a placeholder for now so Python doesn't throw an error
    pass


# 2. Create the actual things (The Objects / Instances)
house_a = House()
house_b = House()

# 3. Prove that they are distinct, unique objects
print(type(house_a))  # Output: <class '__main__.House'>
print(house_a)        # Output: <__main__.House object at 0x1023a4fd0>
print(house_b)        # Output: <__main__.House object at 0x1023a4f50>
```

Adding Data to an Object (The Manual Way)
Right now, our houses are completely empty. In Python, you can dynamically attach data (attributes) directly to an object using dot notation (object.attribute).

```python
class House:
    pass

house_a = House()
house_b = House()

# Assigning unique data to house_a
house_a.color = "Blue"
house_a.bedrooms = 3

# Assigning unique data to house_b
house_b.color = "Yellow"
house_b.bedrooms = 5

# Accessing the data
print(f"House A is {house_a.color} with {house_a.bedrooms} bedrooms.")
print(f"House B is {house_b.color} with {house_b.bedrooms} bedrooms.")
```

The __init__ Method
In the last step, we manually added data to our houses like this: house_a.color = "Blue". As we discussed, that's dangerous because it’s prone to typos and lacks structure.

This is where the __init__ method comes to the rescue.

What is __init__?
The __init__ method is a special function that Python runs automatically the exact millisecond you create a new object. It stands for initialize. It sets up (initializes) the starting data for your object.

In other programming languages (like Java or C++), this is called a Constructor.

Because it has double underscores before and after its name, the Python community calls it a "Dunder" method (Double Under-score). Dunder methods have special superpowers in Python.

The Blueprint Analogy
Think of the __init__ method like a required form you have to fill out at the factory before they will build your custom item.

If you want a car, the factory says: "You cannot have a car until you fill out this form specifying the brand and the model." ---

Enforcement: If you try to create a car without giving it a brand, model, or year, Python will immediately crash and say: TypeError: __init__() missing 3 required positional arguments. It protects you from making incomplete objects.

Consistency: You are guaranteed that every single car object created from this class will have a .brand, a .model, and a .year. No typos, no missing fields.


```python
class Car:
    # The __init__ method defines what data EVERY car must have
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year

# Now, creating a car is a single, clean line of code:
car1 = Car("Toyota", "Camry", 2024)
car2 = Car("Tesla", "Model Y", 2026)

# Accessing the data
print(f"Car 1 is a {car1.year} {car1.brand} {car1.model}.")
print(f"Car 2 is a {car2.year} {car2.brand} {car2.model}.")
```

The self Parameter
What is self?
Simply put, self represents the specific object that is currently calling the method.

When you write a class, you are writing code for objects that don't even exist yet. You need a way to say: "Whichever object ends up using this method later, look at that object's data." Since you don't know if the object's name will be car1, my_ev, or james_truck, Python uses self as a universal placeholder.

The Real-World Analogy
Imagine you are reading the manual for a smartphone. The manual says:

"To change your wallpaper, press the settings button on your screen."

The manual doesn't say "Press the button on John's screen" or "Press the button on Sarah's screen." It uses the word "your" to refer to whichever specific phone is currently in the reader's hand.

In Python, self is exactly like the word "your" or "my".



```python
class Robot:
    def __init__(self, name):
        self.name = name  # "My name is..."

    def introduce(self):
        return f"Hello, my name is {self.name}"

# Create two separate robots
bot1 = Robot("Chappie")
bot2 = Robot("Wall-E")

# <------> #
print(bot1.introduce())  # Output: Hello, my name is Chappie
print(bot2.introduce())  # Output: Hello, my name is Wall-E
```

When you type bot1.introduce(), you aren't passing any arguments inside the parentheses. But notice that in the class definition, introduce(self) expects an argument!

How does this work? Behind the scenes, Python secretly rewrites your code like this:

Python
This is what Python actually executes under the hood:

Robot.introduce(bot1)  bot1 is passed as 'self'

Robot.introduce(bot2)  bot2 is passed as 'self'

Because bot1 is passed in as self, self.name becomes bot1.name, which equals "Chappie".

It must be the first parameter: Every regular method inside a class must have self as its very first argument.

It's not a reserved keyword: You could technically name it this, me, or banana, and Python would still work. However, never do this. The entire Python world uses self. If you use anything else, other programmers (and AI tools) will find your code incredibly hard to read!

There are two major types of attributes you need to know, and confusing them is one of the most common mistakes beginners make.

Instance Attributes (Unique to each object)

Class Attributes (Shared by ALL objects of that class)

1. Instance Attributes
These are variables that belong to a specific object. They are defined inside the __init__ method using self.

If you change an instance attribute on one object, it has absolutely no effect on any other object.

```python
class Dog:
    def __init__(self, name):
        self.name = name  # Instance attribute

dog1 = Dog("Buddy")
dog2 = Dog("Max")

# Changing dog1's name doesn't change dog2's name
dog1.name = "Rocky"

print(dog1.name)  # Output: Rocky
print(dog2.name)  # Output: Max
```

2. Class Attributes
These are variables defined directly inside the class, but outside any methods.

They do not use self. Instead of belonging to an individual object, they belong to the entire class. Every single object created from this blueprint shares this exact same variable and value.

The Real-World Analogy
Think of a human being.

Your name, height, and eye color are Instance Attributes (they are unique to you).

The fact that humans have 2 legs and live on Planet Earth are Class Attributes (it is a universal truth shared by every single human).

```python
class Dog:
    # Class Attribute (Shared by all dogs)
    species = "Canis lupus familiaris"

    def __init__(self, name):
        # Instance Attribute (Unique to each dog)
        self.name = name

# Create two dog objects
dog1 = Dog("Buddy")
dog2 = Dog("Max")

# Both dogs share the same species
print(dog1.species)  # Output: Canis lupus familiaris
print(dog2.species)  # Output: Canis lupus familiaris
```

If you change a Class Attribute by targeting the class itself, every single object changes instantly.
Use them for data that should remain constant across all instances, such as:

Constant configurations (e.g., MAX_SPEED = 120 for a specific video game character class).

Tracker counters (e.g., keeping track of how many total users have registered in a system).

```python
# Let's update the species for the entire Dog class
Dog.species = "Canines"

# Now check both dogs again:
print(dog1.species)  # Output: Canines
print(dog2.species)  # Output: Canines
```

```python
class User:
    user_count = 0  # Class attribute to track total users

    def __init__(self, username):
        self.username = username
        # Every time a user is created, increment the global class counter
        User.user_count += 1

u1 = User("alice")
u2 = User("bob")

print(User.user_count)  # Output: 2
```

1. Instance Methods
These are the standard methods you write 90% of the time.

They always take self as their first parameter.

They can read and modify data unique to a specific object instance.

Why it's an instance method: It needs to know about self.size and self.amount_left. If you drink from your cup, my cup shouldn't become emptier!

```python
class CoffeeCup:
    def __init__(self, size):
        self.size = size          # Instance attribute
        self.amount_left = 100    # Instance attribute (percentage)

    # This is an Instance Method
    def take_sip(self):
        self.amount_left -= 10
        return f"Delicious! Your {self.size} cup is now {self.amount_left}% full."

my_cup = CoffeeCup("Large")
print(my_cup.take_sip())  # Output: Delicious! Your Large cup is now 90% full.
```

2. Class Methods
Sometimes, a method doesn't care about individual objects. It needs to look at or change something for the entire class.

To make a class method, we do two things:

We put a special decorator above it: @classmethod.

Instead of self, we pass cls as the first parameter. cls stands for "Class".

The Code Example
Let's look at a pizza shop. The price of a pizza depends on inflation, which affects all pizzas equally.

```python
class Pizza:
    base_price = 10  # Class attribute: Base price for ALL pizzas

    def __init__(self, topping):
        self.topping = topping

    # A Class Method to adjust the price for the whole business
    @classmethod
    def adjust_for_inflation(cls, extra_cost):
        cls.base_price += extra_cost  # Modifying the Class Attribute using 'cls'

    def get_price(self):
        return f"Your {self.topping} pizza costs ${Pizza.base_price}"

# Create some pizzas
p1 = Pizza("Pepperoni")
p2 = Pizza("Mushroom")

print(p1.get_price())  # Output: Your Pepperoni pizza costs $10

# Suddenly, inflation hits! We update the class, not the individual pizzas.
Pizza.adjust_for_inflation(3)

print(p1.get_price())  # Output: Your Pepperoni pizza costs $13
print(p2.get_price())  # Output: Your Mushroom pizza costs $13
```
